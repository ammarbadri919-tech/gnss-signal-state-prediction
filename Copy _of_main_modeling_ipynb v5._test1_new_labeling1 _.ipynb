```python
# ============================================================
# REGENERATE ALL 196 FILES WITH CORRECTED 20 FEATURES
# ============================================================

import os
import zipfile
import pandas as pd
import numpy as np

zip_path = "/content/drive/MyDrive/Gnns_kaggle_2324/smartphone-decimeter-2023.zip"
corrected_dir = "/content/gnss_features_v2"

os.makedirs(corrected_dir, exist_ok=True)

# ------------------------------------------------------------
# 20 MODEL FEATURES
# ------------------------------------------------------------

input_features = [
    "Cn0DbHz",
    "Cn0_change",
    "Cn0_trend",
    "Cn0_trend_mean",
    "down_streak",
    "up_streak",
    "PseudorangeRateMetersPerSecond",
    "PseudorangeRate_change",
    "SvElevationDegrees",
    "Elevation_change",
    "SvAzimuthDegrees",
    "Azimuth_change",
    "Cn0_rolling_std",
    "Cn0_slope",
    "Cn0_acceleration",
    "Cn0_min_5",
    "Cn0_max_5",
    "Cn0_range_5",
    "negative_change_count_10",
    "positive_change_count_10"
]

# ------------------------------------------------------------
# FIND ALL 196 GNSS FILES
# ------------------------------------------------------------

with zipfile.ZipFile(zip_path, "r") as z:

    gnss_files = sorted([
        f for f in z.namelist()
        if f.endswith("/device_gnss.csv")
    ])

print("Total GNSS files:", len(gnss_files))

assert len(gnss_files) == 196, \
    f"Expected 196 files, found {len(gnss_files)}"

# ------------------------------------------------------------
# PROCESS EACH FILE
# ------------------------------------------------------------

for idx, file_path in enumerate(gnss_files, start=1):

    print(f"\n[{idx}/196] Processing:")
    print(file_path)

    # --------------------------------------------------------
    # READ ONE FILE ONLY
    # --------------------------------------------------------

    with zipfile.ZipFile(zip_path, "r") as z:
        df = pd.read_csv(z.open(file_path))

    # --------------------------------------------------------
    # SELECT REQUIRED RAW FEATURES
    # --------------------------------------------------------

    raw_features = [
        "ConstellationType",
        "Svid",
        "SignalType",
        "utcTimeMillis",
        "Cn0DbHz",
        "RawPseudorangeMeters",
        "PseudorangeRateMetersPerSecond",
        "AccumulatedDeltaRangeMeters",
        "SvElevationDegrees",
        "SvAzimuthDegrees"
    ]

    df = df[raw_features].copy()

    # --------------------------------------------------------
    # CREATE SIGNAL ID
    # --------------------------------------------------------

    df["signal_id"] = (
        df["ConstellationType"].astype(str)
        + "_"
        + df["Svid"].astype(str)
        + "_"
        + df["SignalType"].astype(str)
    )

    # --------------------------------------------------------
    # SORT BY SIGNAL + TIME
    # --------------------------------------------------------

    df = df.sort_values(
        ["signal_id", "utcTimeMillis"]
    ).reset_index(drop=True)

    # --------------------------------------------------------
    # INTERPOLATION
    # --------------------------------------------------------

    numeric_cols = [
        "Cn0DbHz",
        "RawPseudorangeMeters",
        "PseudorangeRateMetersPerSecond",
        "AccumulatedDeltaRangeMeters",
        "SvElevationDegrees",
        "SvAzimuthDegrees"
    ]

    df[numeric_cols] = (
        df.groupby("signal_id")[numeric_cols]
          .transform(
              lambda x: x.interpolate(
                  method="linear",
                  limit=2,
                  limit_direction="both"
              )
          )
    )

    df = df.dropna(
        subset=numeric_cols
    ).reset_index(drop=True)

    # --------------------------------------------------------
    # C/N0 CHANGE
    # --------------------------------------------------------

    df["Cn0_change"] = (
        df.groupby("signal_id")["Cn0DbHz"].diff()
    )

    # --------------------------------------------------------
    # C/N0 ROLLING MEAN
    # --------------------------------------------------------

    df["Cn0_rolling_mean"] = (
        df.groupby("signal_id")["Cn0DbHz"]
          .transform(
              lambda x: x.rolling(
                  window=5,
                  min_periods=1
              ).mean()
          )
    )

    # --------------------------------------------------------
    # C/N0 TREND
    # --------------------------------------------------------

    df["Cn0_trend"] = (
        df.groupby("signal_id")["Cn0DbHz"]
          .transform(
              lambda x: x.diff(4)
          )
    )

    # --------------------------------------------------------
    # C/N0 TREND MEAN
    # --------------------------------------------------------

    df["Cn0_trend_mean"] = (
        df.groupby("signal_id")["Cn0_change"]
          .transform(
              lambda x: x.rolling(
                  window=3,
                  min_periods=1
              ).mean()
          )
    )

    # --------------------------------------------------------
    # UP / DOWN
    # --------------------------------------------------------

    df["is_up"] = (
        df["Cn0_trend_mean"] > 0
    )

    df["up_streak"] = (
        df.groupby("signal_id")["is_up"]
          .transform(
              lambda x:
                  x.astype(int)
                   .groupby((x != x.shift()).cumsum())
                   .cumsum()
          )
    )

    df["is_down"] = (
        df["Cn0_trend_mean"] < 0
    )

    df["down_streak"] = (
        df.groupby("signal_id")["is_down"]
          .transform(
              lambda x:
                  x.astype(int)
                   .groupby((x != x.shift()).cumsum())
                   .cumsum()
          )
    )

    # --------------------------------------------------------
    # NEW FEATURE 1: ROLLING STD
    # --------------------------------------------------------

    df["Cn0_rolling_std"] = (
        df.groupby("signal_id")["Cn0DbHz"]
          .transform(
              lambda x: x.rolling(
                  window=5,
                  min_periods=2
              ).std()
          )
    )

    # --------------------------------------------------------
    # NEW FEATURE 2: SLOPE
    # --------------------------------------------------------

    df["Cn0_slope"] = (
        df.groupby("signal_id")["Cn0DbHz"]
          .transform(
              lambda x: x.diff(4) / 4
          )
    )

    # --------------------------------------------------------
    # NEW FEATURE 3: ACCELERATION
    # --------------------------------------------------------

    df["Cn0_acceleration"] = (
        df.groupby("signal_id")["Cn0_slope"].diff()
    )

    # --------------------------------------------------------
    # TIME GAP
    # --------------------------------------------------------

    df["time_gap"] = (
        df.groupby("signal_id")["utcTimeMillis"].diff()
    )

    # --------------------------------------------------------
    # PSEUDORANGE CHANGE
    # --------------------------------------------------------

    df["Pseudorange_change"] = (
        df.groupby("signal_id")[
            "RawPseudorangeMeters"
        ].diff()
    )

    # --------------------------------------------------------
    # PSEUDORANGE RATE CHANGE
    # --------------------------------------------------------

    df["PseudorangeRate_change"] = (
        df.groupby("signal_id")[
            "PseudorangeRateMetersPerSecond"
        ].diff()
    )

    # --------------------------------------------------------
    # ELEVATION CHANGE
    # --------------------------------------------------------

    df["Elevation_change"] = (
        df.groupby("signal_id")[
            "SvElevationDegrees"
        ].diff()
    )

    # --------------------------------------------------------
    # CIRCULAR AZIMUTH CHANGE
    # --------------------------------------------------------

    df["Azimuth_change"] = (
        (
            df.groupby("signal_id")[
                "SvAzimuthDegrees"
            ].diff() + 180
        ) % 360
    ) - 180

    df["Azimuth_change"] = (
        df["Azimuth_change"].clip(-180, 180)
    )

    # --------------------------------------------------------
    # NEW FEATURE 4: MIN
    # --------------------------------------------------------

    df["Cn0_min_5"] = (
        df.groupby("signal_id")["Cn0DbHz"]
          .transform(
              lambda x: x.rolling(
                  window=5,
                  min_periods=1
              ).min()
          )
    )

    # --------------------------------------------------------
    # NEW FEATURE 5: MAX
    # --------------------------------------------------------

    df["Cn0_max_5"] = (
        df.groupby("signal_id")["Cn0DbHz"]
          .transform(
              lambda x: x.rolling(
                  window=5,
                  min_periods=1
              ).max()
          )
    )

    # --------------------------------------------------------
    # NEW FEATURE 6: RANGE
    # --------------------------------------------------------

    df["Cn0_range_5"] = (
        df["Cn0_max_5"] - df["Cn0_min_5"]
    )

    # --------------------------------------------------------
    # NEW FEATURE 7: NEGATIVE COUNT
    # --------------------------------------------------------

    df["negative_change_count_10"] = (
        df.groupby("signal_id")["Cn0_change"]
          .transform(
              lambda x:
                  (x < 0)
                  .rolling(
                      window=10,
                      min_periods=1
                  )
                  .sum()
          )
    )

    # --------------------------------------------------------
    # NEW FEATURE 8: POSITIVE COUNT
    # --------------------------------------------------------

    df["positive_change_count_10"] = (
        df.groupby("signal_id")["Cn0_change"]
          .transform(
              lambda x:
                  (x > 0)
                  .rolling(
                      window=10,
                      min_periods=1
                  )
                  .sum()
          )
    )

    # REMOVE ROWS WHERE 20 FEATURES ARE INCOMPLETE

    df = df.dropna(
        subset=input_features
    ).reset_index(drop=True)


    # VERIFY

    missing_features = [
        f for f in input_features
        if f not in df.columns
    ]

    assert not missing_features, \
        f"Missing features: {missing_features}"

    remaining_nan = (
        df[input_features]
        .isna()
        .sum()
        .sum()
    )

    assert remaining_nan == 0, \
        f"Remaining NaNs: {remaining_nan}"

    output_name = f"gnss_{idx:03d}.parquet"

    output_path = os.path.join(
        corrected_dir,
        output_name
    )

    df.to_parquet(
        output_path,
        index=False
    )

    print(
        f"Saved {output_name} | "
        f"Rows: {len(df)} | "
        f"Columns: {len(df.columns)}"
    )

print("\n========================================")
print("DONE — ALL 196 FILES PROCESSED")
print("========================================")
```

```python

from pathlib import Path
import numpy as np
import pandas as pd
import pyarrow.parquet as pq

corrected_dir = Path("/content/gnss_features_v2")

analysis_features = [
    "Cn0DbHz",
    "Cn0_trend",
    "Cn0_trend_mean",
    "Cn0_slope",
    "Cn0_acceleration",
    "Cn0_rolling_std",
    "Cn0_min_5",
    "Cn0_max_5",
    "Cn0_range_5",
    "negative_change_count_10",
    "positive_change_count_10",
    "up_streak",
    "down_streak",
]

sample_size_per_feature = 50_000
batch_size = 32_768
rng = np.random.default_rng(42)

reservoirs = {
    feature: []
    for feature in analysis_features
}

for i in range(1, 138):

    path = corrected_dir / f"gnss_{i:03d}.parquet"

    parquet_file = pq.ParquetFile(path)

    try:
        for batch in parquet_file.iter_batches(
            batch_size=batch_size,
            columns=analysis_features,
            use_threads=False
        ):

            for feature in analysis_features:

                column = batch.column(
                    batch.schema.get_field_index(feature)
                )

                values = np.asarray(
                    column.to_numpy(
                        zero_copy_only=False
                    ),
                    dtype=np.float64
                )

                values = values[
                    np.isfinite(values)
                ]

                if values.size == 0:
                    continue

                reservoirs[feature].extend(
                    values.tolist()
                )

                # Keep memory bounded
                if len(reservoirs[feature]) > sample_size_per_feature:

                    selected = rng.choice(
                        len(reservoirs[feature]),
                        size=sample_size_per_feature,
                        replace=False
                    )

                    reservoirs[feature] = [
                        reservoirs[feature][j]
                        for j in selected
                    ]

    finally:
        parquet_file.close()

    print(f"[{i}/137] processed")


# PERCENTILE TABLE

percentile_table = pd.DataFrame(
    index=analysis_features,
    columns=[
        "p10",
        "p25",
        "p50",
        "p75",
        "p90",
        "p95"
    ],
    dtype=float
)

for feature in analysis_features:

    values = np.asarray(
        reservoirs[feature],
        dtype=np.float64
    )

    percentile_table.loc[feature] = np.percentile(
        values,
        [10, 25, 50, 75, 90, 95]
    )


print("TRAINING FEATURE DISTRIBUTIONS")

display(
    percentile_table.round(4)
)
```

```python
slope_results = []

for i in range(1, 138):

    path = corrected_dir / f"gnss_{i:03d}.parquet"

    df = pd.read_parquet(
        path,
        columns=["Cn0DbHz", "Cn0_slope"]
    )

    df["region"] = pd.cut(
        df["Cn0DbHz"],
        bins=[-np.inf, p25, p50, p75, np.inf],
        labels=["LOW", "LOW_MID", "HIGH_MID", "HIGH"],
        include_lowest=True
    )

    grouped = (
        df.groupby("region", observed=False)
        .agg(
            count=("Cn0_slope", "count"),
            mean_slope=("Cn0_slope", "mean")
        )
        .reset_index()
    )

    slope_results.append(grouped)

slope_all = (
    pd.concat(slope_results, ignore_index=True)
    .groupby("region", observed=False)
    .agg(
        count=("count", "sum")
    )
    .reset_index()
)

# Calculate weighted mean properly
weighted_sum = (
    pd.concat(slope_results, ignore_index=True)
    .assign(
        slope_sum=lambda x:
        x["count"] * x["mean_slope"]
    )
    .groupby("region", observed=False)["slope_sum"]
    .sum()
)

slope_all["mean_slope"] = (
    weighted_sum.values /
    slope_all["count"].values
)

display(
    slope_all[
        ["region", "count", "mean_slope"]
    ].round(4)
)
```

```python
import numpy as np
import pandas as pd

all_features = [
    "Cn0DbHz",
    "Cn0_change",
    "Cn0_trend",
    "Cn0_trend_mean",
    "down_streak",
    "up_streak",
    "PseudorangeRateMetersPerSecond",
    "PseudorangeRate_change",
    "SvElevationDegrees",
    "Elevation_change",
    "SvAzimuthDegrees",
    "Azimuth_change",
    "Cn0_rolling_std",
    "Cn0_slope",
    "Cn0_acceleration",
    "Cn0_min_5",
    "Cn0_max_5",
    "Cn0_range_5",
    "negative_change_count_10",
    "positive_change_count_10"
]

regions = ["LOW", "LOW_MID", "HIGH_MID", "HIGH"]

# ---------------------------------------------------------
# Storage
# ---------------------------------------------------------

stats = {}

for feature in all_features:
    for region in regions:
        stats[(feature, region)] = {
            "count": 0,
            "sum": 0.0,
            "sum_sq": 0.0,
            "min": np.inf,
            "max": -np.inf,
            "negative_count": 0,
            "positive_count": 0
        }

# ---------------------------------------------------------
# Process all 137 training files
# ---------------------------------------------------------

for i in range(1, 138):

    path = corrected_dir / f"gnss_{i:03d}.parquet"

    df = pd.read_parquet(
        path,
        columns=all_features
    )

    # C/N0 defines the regions
    valid_cn0 = np.isfinite(df["Cn0DbHz"])

    df = df.loc[valid_cn0].copy()

    df["region"] = pd.cut(
        df["Cn0DbHz"],
        bins=[-np.inf, p25, p50, p75, np.inf],
        labels=regions,
        include_lowest=True
    )

    # -----------------------------------------------------
    # Collect statistics for every feature in every region
    # -----------------------------------------------------

    for region in regions:

        region_df = df[df["region"] == region]

        if region_df.empty:
            continue

        for feature in all_features:

            values = region_df[feature].to_numpy(
                dtype=np.float64
            )

            values = values[np.isfinite(values)]

            if values.size == 0:
                continue

            s = stats[(feature, region)]

            s["count"] += values.size
            s["sum"] += values.sum()
            s["sum_sq"] += np.square(values).sum()
            s["min"] = min(s["min"], values.min())
            s["max"] = max(s["max"], values.max())

            s["negative_count"] += int(
                (values < 0).sum()
            )

            s["positive_count"] += int(
                (values > 0).sum()
            )

    if i % 10 == 0 or i == 137:
        print(f"[{i}/137] processed")


# ---------------------------------------------------------
# Build ONE combined 80-row table
# ---------------------------------------------------------

rows = []

for feature in all_features:

    for region in regions:

        s = stats[(feature, region)]
        count = s["count"]

        if count == 0:
            continue

        mean = s["sum"] / count

        variance = (
            s["sum_sq"] / count
            - mean ** 2
        )

        variance = max(variance, 0.0)

        std = np.sqrt(variance)

        rows.append({
            "region": region,
            "feature": feature,
            "count": count,
            "mean": mean,
            "std": std,
            "min": s["min"],
            "max": s["max"],
            "negative_pct": (
                s["negative_count"] / count * 100
            ),
            "positive_pct": (
                s["positive_count"] / count * 100
            )
        })

master_behavior_table = pd.DataFrame(rows)

print("\n==============================================")
print("COMBINED MULTI-FEATURE BEHAVIOR TABLE")
print("TRAINING FILES 001–137")
print("==============================================")

print(
    "Rows in final table:",
    len(master_behavior_table)
)

display(
    master_behavior_table.round(4)
)
```

```python

results = []

for i in range(1, 138):

    path = corrected_dir / f"gnss_{i:03d}.parquet"

    df = pd.read_parquet(
        path,
        columns=remaining_features
    )

    df["region"] = pd.cut(
        df["Cn0DbHz"],
        bins=[-np.inf, p25, p50, p75, np.inf],
        labels=region_labels,
        include_lowest=True
    )

    for region in region_labels:

        region_df = df[df["region"] == region]

        if len(region_df) == 0:
            continue

        for feature in remaining_features:

            values = region_df[feature].to_numpy(
                dtype=np.float64
            )

            values = values[np.isfinite(values)]

            if len(values) == 0:
                continue

            results.append({
                "region": region,
                "feature": feature,
                "count": len(values),
                "mean": values.mean(),
                "std": values.std(),
                "min": values.min(),
                "max": values.max(),
                "negative_pct": (values < 0).mean() * 100,
                "positive_pct": (values > 0).mean() * 100
            })

    if i % 10 == 0 or i == 137:
        print(f"[{i}/137] processed")

master_table = pd.DataFrame(results)

print("ALL REMAINING FEATURES — TRAINING DATA")

display(
    master_table.round(4)
)
```

```python
# ==========================================
# STEP 1: LABEL EACH FEATURE
# ==========================================

def label_feature(value, p25, p50, p75, reverse=False):
    if reverse:
        # For negative-direction features
        if value >= p75:
            return "BAD"
        elif value >= p50:
            return "DEGRADING"
        elif value >= p25:
            return "RECOVERY"
        else:
            return "GOOD"

    else:
        if value <= p25:
            return "BAD"
        elif value <= p50:
            return "DEGRADING"
        elif value < p75:
            return "RECOVERY"
        else:
            return "GOOD"
```

```python
# ==========================================
# APPLY FEATURE LABELS
# ==========================================

def add_feature_labels(df):

    # Signal strength
    df["Cn0DbHz_state"] = df["Cn0DbHz"].apply(
        lambda x: label_feature(
            x,
            thresholds["Cn0DbHz"]["P25"],
            thresholds["Cn0DbHz"]["P50"],
            thresholds["Cn0DbHz"]["P75"]
        )
    )

    df["Cn0_min_5_state"] = df["Cn0_min_5"].apply(
        lambda x: label_feature(
            x,
            thresholds["Cn0_min_5"]["P25"],
            thresholds["Cn0_min_5"]["P50"],
            thresholds["Cn0_min_5"]["P75"]
        )
    )

    df["Cn0_max_5_state"] = df["Cn0_max_5"].apply(
        lambda x: label_feature(
            x,
            thresholds["Cn0_max_5"]["P25"],
            thresholds["Cn0_max_5"]["P50"],
            thresholds["Cn0_max_5"]["P75"]
        )
    )

    # Direction
    for feature in [
        "Cn0_change",
        "Cn0_trend",
        "Cn0_trend_mean",
        "Cn0_slope"
    ]:
        df[f"{feature}_state"] = df[feature].apply(
            lambda x, f=feature: label_feature(
                x,
                thresholds[f]["P25"],
                thresholds[f]["P50"],
                thresholds[f]["P75"]
            )
        )

    # Negative persistence: high = BAD
    f = "negative_change_count_10"
    df[f"{f}_state"] = df[f].apply(
        lambda x: label_feature(
            x,
            thresholds[f]["P25"],
            thresholds[f]["P50"],
            thresholds[f]["P75"],
            reverse=True
        )
    )

    # Positive persistence: low = BAD, high = GOOD
    f = "positive_change_count_10"
    df[f"{f}_state"] = df[f].apply(
        lambda x: label_feature(
            x,
            thresholds[f]["P25"],
            thresholds[f]["P50"],
            thresholds[f]["P75"]
        )
    )

    # Instability: high = BAD
    for feature in ["Cn0_rolling_std", "Cn0_range_5"]:
        df[f"{feature}_state"] = df[feature].apply(
            lambda x, f=feature: label_feature(
                x,
                thresholds[f]["P25"],
                thresholds[f]["P50"],
                thresholds[f]["P75"],
                reverse=True
            )
        )

    return df
```

```python
# ==========================================
# FEATURE-LEVEL STATE LABELING
# ==========================================

def label_normal(value, p25, p50, p75):
    if value <= p25:
        return "BAD"
    elif value <= p50:
        return "DEGRADING"
    elif value < p75:
        return "RECOVERY"
    else:
        return "GOOD"


def label_reverse(value, p25, p50, p75):
    # High value = BAD
    if value >= p75:
        return "BAD"
    elif value >= p50:
        return "DEGRADING"
    elif value >= p25:
        return "RECOVERY"
    else:
        return "GOOD"


def label_features(df):

    # Signal strength
    for f in [
        "Cn0DbHz",
        "Cn0_min_5",
        "Cn0_max_5"
    ]:
        df[f + "_state"] = df[f].apply(
            lambda x, f=f: label_normal(
                x,
                thresholds[f]["P25"],
                thresholds[f]["P50"],
                thresholds[f]["P75"]
            )
        )

    # Direction
    for f in [
        "Cn0_change",
        "Cn0_trend",
        "Cn0_trend_mean",
        "Cn0_slope"
    ]:
        df[f + "_state"] = df[f].apply(
            lambda x, f=f: label_normal(
                x,
                thresholds[f]["P25"],
                thresholds[f]["P50"],
                thresholds[f]["P75"]
            )
        )

    # Negative persistence: high = BAD
    f = "negative_change_count_10"
    df[f + "_state"] = df[f].apply(
        lambda x: label_reverse(
            x,
            thresholds[f]["P25"],
            thresholds[f]["P50"],
            thresholds[f]["P75"]
        )
    )

    # Positive persistence: low = BAD, high = GOOD
    f = "positive_change_count_10"
    df[f + "_state"] = df[f].apply(
        lambda x: label_normal(
            x,
            thresholds[f]["P25"],
            thresholds[f]["P50"],
            thresholds[f]["P75"]
        )
    )

    # Instability: high = BAD
    for f in [
        "Cn0_rolling_std",
        "Cn0_range_5"
    ]:
        df[f + "_state"] = df[f].apply(
            lambda x, f=f: label_reverse(
                x,
                thresholds[f]["P25"],
                thresholds[f]["P50"],
                thresholds[f]["P75"]
            )
        )

    return df
```

```python
# ==========================================
# FINAL DIRECT LABELING RULE
# ==========================================

def get_final_state(row):

    # -------------------------
    # Thresholds
    # -------------------------
    cn0_p25 = thresholds["Cn0DbHz"]["P25"]
    cn0_p50 = thresholds["Cn0DbHz"]["P50"]
    cn0_p75 = thresholds["Cn0DbHz"]["P75"]

    min_p25 = thresholds["Cn0_min_5"]["P25"]
    max_p75 = thresholds["Cn0_max_5"]["P75"]

    trend_p25 = thresholds["Cn0_trend"]["P25"]

    neg_p50 = thresholds["negative_change_count_10"]["P50"]
    neg_p75 = thresholds["negative_change_count_10"]["P75"]

    pos_p50 = thresholds["positive_change_count_10"]["P50"]

    std_p50 = thresholds["Cn0_rolling_std"]["P50"]

    range_p75 = thresholds["Cn0_range_5"]["P75"]

    # -------------------------
    # Current values
    # -------------------------
    cn0 = row["Cn0DbHz"]
    cn0_min = row["Cn0_min_5"]
    cn0_max = row["Cn0_max_5"]

    trend = row["Cn0_trend"]

    neg_count = row["negative_change_count_10"]
    pos_count = row["positive_change_count_10"]

    rolling_std = row["Cn0_rolling_std"]
    range_5 = row["Cn0_range_5"]

    # ==========================================
    # 1. BAD
    # ==========================================

    if (
        cn0 <= cn0_p25
        and cn0_min <= min_p25
        and trend <= trend_p25
        and neg_count >= neg_p75
        and (
            rolling_std >= std_p50
            or range_5 >= range_p75
        )
    ):
        return "BAD"

    # ==========================================
    # 2. LOW SIGNAL
    # ==========================================

    if cn0 <= cn0_p25:

        # Strong negative behavior
        if (
            trend < 0
            and neg_count >= neg_p50
        ):
            return "DEGRADING"

        # Signal is low but moving upward
        return "RECOVERY"

    # ==========================================
    # 3. LOW-MEDIUM SIGNAL
    # ==========================================

    if cn0 <= cn0_p50:

        if trend < 0:
            return "DEGRADING"

        return "RECOVERY"

    # ==========================================
    # 4. MEDIUM-HIGH SIGNAL
    # ==========================================

    if cn0 < cn0_p75:

        if trend < 0:
            return "DEGRADING"

        return "RECOVERY"

    # ==========================================
    # 5. HIGH SIGNAL
    # ==========================================

    if cn0 >= cn0_p75:

        # Stable strong signal
        if (
            trend >= 0
            and cn0_max >= max_p75
            and rolling_std <= std_p50
            and pos_count >= pos_p50
        ):
            return "GOOD"

        # High signal but unstable / changing
        if trend < 0:
            return "DEGRADING"

        return "RECOVERY"

    return "RECOVERY"
```

```python
# ==========================================
# LABEL ALL 196 GNSS FILES
# ==========================================

import os
import numpy as np
import pandas as pd

INPUT_DIR = "/content/gnss_features_v2"
OUTPUT_DIR = "/content/gnss_labeled_final"

os.makedirs(OUTPUT_DIR, exist_ok=True)


def final_state_vectorized(df):

    cn0 = df["Cn0DbHz"].to_numpy()
    cn0_min = df["Cn0_min_5"].to_numpy()
    cn0_max = df["Cn0_max_5"].to_numpy()

    trend = df["Cn0_trend"].to_numpy()

    neg_count = df["negative_change_count_10"].to_numpy()
    pos_count = df["positive_change_count_10"].to_numpy()

    rolling_std = df["Cn0_rolling_std"].to_numpy()
    range_5 = df["Cn0_range_5"].to_numpy()

    # Thresholds
    cn0_p25 = thresholds["Cn0DbHz"]["P25"]
    cn0_p50 = thresholds["Cn0DbHz"]["P50"]
    cn0_p75 = thresholds["Cn0DbHz"]["P75"]

    min_p25 = thresholds["Cn0_min_5"]["P25"]
    max_p75 = thresholds["Cn0_max_5"]["P75"]

    trend_p25 = thresholds["Cn0_trend"]["P25"]

    neg_p50 = thresholds["negative_change_count_10"]["P50"]
    neg_p75 = thresholds["negative_change_count_10"]["P75"]

    pos_p50 = thresholds["positive_change_count_10"]["P50"]

    std_p50 = thresholds["Cn0_rolling_std"]["P50"]
    range_p75 = thresholds["Cn0_range_5"]["P75"]

    # Default
    state = np.full(
        len(df),
        "RECOVERY",
        dtype=object
    )

    # BAD
    bad = (
        (cn0 <= cn0_p25) &
        (cn0_min <= min_p25) &
        (trend <= trend_p25) &
        (neg_count >= neg_p75) &
        (
            (rolling_std >= std_p50) |
            (range_5 >= range_p75)
        )
    )

    state[bad] = "BAD"

    # DEGRADING
    degrading_low = (
        (~bad) &
        (cn0 <= cn0_p25) &
        (trend < 0) &
        (neg_count >= neg_p50)
    )

    degrading_mid = (
        (~bad) &
        (cn0 > cn0_p25) &
        (cn0 <= cn0_p75) &
        (trend < 0)
    )

    degrading_high = (
        (~bad) &
        (cn0 > cn0_p75) &
        (trend < 0)
    )

    degrading = (
        degrading_low |
        degrading_mid |
        degrading_high
    )

    state[degrading] = "DEGRADING"

    # GOOD
    good = (
        (~bad) &
        (~degrading) &
        (cn0 >= cn0_p75) &
        (trend >= 0) &
        (cn0_max >= max_p75) &
        (rolling_std <= std_p50) &
        (pos_count >= pos_p50)
    )

    state[good] = "GOOD"

    return state


# ------------------------------------------
# PROCESS ALL 196 FILES
# ------------------------------------------

total_rows = 0

for i in range(1, 197):

    input_path = (
        f"{INPUT_DIR}/gnss_{i:03d}.parquet"
    )

    output_path = (
        f"{OUTPUT_DIR}/gnss_{i:03d}.parquet"
    )

    df = pd.read_parquet(input_path)

    df["state"] = final_state_vectorized(df)

    df.to_parquet(
        output_path,
        index=False
    )

    total_rows += len(df)

    print(
        f"[{i}/196] "
        f"{len(df):,} rows labeled"
    )

print("\n================================")
print("ALL FILES LABELED")
print("================================")
print("Total rows:", f"{total_rows:,}")
print("Output:", OUTPUT_DIR)
```

```python
# ==========================================
# SEQUENCE DATASET CONFIGURATION
# ==========================================

import tensorflow as tf
import pandas as pd
import numpy as np

DATA_DIR = "/content/gnss_labeled_final"

SEQUENCE_LENGTH = 20
PREDICTION_HORIZON = 5
BATCH_SIZE = 256

input_features = [
    "Cn0DbHz",
    "Cn0_change",
    "Cn0_trend",
    "Cn0_trend_mean",
    "down_streak",
    "up_streak",
    "PseudorangeRateMetersPerSecond",
    "PseudorangeRate_change",
    "SvElevationDegrees",
    "Elevation_change",
    "SvAzimuthDegrees",
    "Azimuth_change",
    "Cn0_rolling_std",
    "Cn0_slope",
    "Cn0_acceleration",
    "Cn0_min_5",
    "Cn0_max_5",
    "Cn0_range_5",
    "negative_change_count_10",
    "positive_change_count_10"
]

print("Data directory:", DATA_DIR)
print("Sequence length:", SEQUENCE_LENGTH)
print("Prediction horizon:", PREDICTION_HORIZON)
print("Number of features:", len(input_features))
```

```python
# ==========================================
# STATE → NUMERIC LABEL
# ==========================================

state_to_number = {
    "BAD": 0,
    "DEGRADING": 1,
    "RECOVERY": 2,
    "GOOD": 3
}

number_to_state = {
    v: k for k, v in state_to_number.items()
}

print("Label mapping:")

for state, number in state_to_number.items():
    print(f"{state} -> {number}")
```

```python
# ==========================================
# RAM-SAFE SEQUENCE GENERATOR
# ==========================================

def batch_generator(start_file, end_file):

    X_batch = []
    y_batch = []

    for file_num in range(start_file, end_file + 1):

        file_path = (
            f"{DATA_DIR}/gnss_{file_num:03d}.parquet"
        )

        data = pd.read_parquet(
            file_path,
            columns=input_features + [
                "signal_id",
                "utcTimeMillis",
                "state"
            ]
        )

        for signal_id, signal_df in data.groupby(
            "signal_id",
            sort=False
        ):

            signal_df = (
                signal_df
                .sort_values("utcTimeMillis")
                .reset_index(drop=True)
            )

            features = signal_df[
                input_features
            ].to_numpy(dtype=np.float32)

            labels = np.array([
                state_to_number[state]
                for state in signal_df["state"]
            ], dtype=np.int32)

            max_start = (
                len(signal_df)
                - SEQUENCE_LENGTH
                - PREDICTION_HORIZON
                + 1
            )

            for i in range(max_start):

                X_batch.append(
                    features[
                        i:i + SEQUENCE_LENGTH
                    ]
                )

                target_index = (
                    i
                    + SEQUENCE_LENGTH
                    + PREDICTION_HORIZON
                    - 1
                )

                y_batch.append(
                    labels[target_index]
                )

                if len(X_batch) == BATCH_SIZE:

                    yield (
                        np.asarray(
                            X_batch,
                            dtype=np.float32
                        ),
                        np.asarray(
                            y_batch,
                            dtype=np.int32
                        )
                    )

                    X_batch = []
                    y_batch = []

    if X_batch:

        yield (
            np.asarray(
                X_batch,
                dtype=np.float32
            ),
            np.asarray(
                y_batch,
                dtype=np.int32
            )
        )


print("Sequence generator created.")
```
