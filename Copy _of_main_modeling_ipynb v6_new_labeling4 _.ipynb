```python
# ============================================================
# REGENERATE ALL 196 FILES WITH CORRECTED 20 FEATURES
# ============================================================

import os
import zipfile
import pandas as pd
import numpy as np

zip_path = "/content/drive/MyDrive/Gnns_kaggle_2324/smartphone-decimeter-2023.zip"
corrected_dir = "/content/gnss_features_v2"

os.makedirs(corrected_dir, exist_ok=True)

# ------------------------------------------------------------
# 20 MODEL FEATURES
# ------------------------------------------------------------

input_features = [
    "Cn0DbHz",
    "Cn0_change",
    "Cn0_trend",
    "Cn0_trend_mean",
    "down_streak",
    "up_streak",
    "PseudorangeRateMetersPerSecond",
    "PseudorangeRate_change",
    "SvElevationDegrees",
    "Elevation_change",
    "SvAzimuthDegrees",
    "Azimuth_change",
    "Cn0_rolling_std",
    "Cn0_slope",
    "Cn0_acceleration",
    "Cn0_min_5",
    "Cn0_max_5",
    "Cn0_range_5",
    "negative_change_count_10",
    "positive_change_count_10"
]

# ------------------------------------------------------------
# FIND ALL 196 GNSS FILES
# ------------------------------------------------------------

with zipfile.ZipFile(zip_path, "r") as z:

    gnss_files = sorted([
        f for f in z.namelist()
        if f.endswith("/device_gnss.csv")
    ])

print("Total GNSS files:", len(gnss_files))

assert len(gnss_files) == 196, \
    f"Expected 196 files, found {len(gnss_files)}"

# ------------------------------------------------------------
# PROCESS EACH FILE
# ------------------------------------------------------------

for idx, file_path in enumerate(gnss_files, start=1):

    print(f"\n[{idx}/196] Processing:")
    print(file_path)

    # --------------------------------------------------------
    # READ ONE FILE ONLY
    # --------------------------------------------------------

    with zipfile.ZipFile(zip_path, "r") as z:
        df = pd.read_csv(z.open(file_path))

    # --------------------------------------------------------
    # SELECT REQUIRED RAW FEATURES
    # --------------------------------------------------------

    raw_features = [
        "ConstellationType",
        "Svid",
        "SignalType",
        "utcTimeMillis",
        "Cn0DbHz",
        "RawPseudorangeMeters",
        "PseudorangeRateMetersPerSecond",
        "AccumulatedDeltaRangeMeters",
        "SvElevationDegrees",
        "SvAzimuthDegrees"
    ]

    df = df[raw_features].copy()

    # --------------------------------------------------------
    # CREATE SIGNAL ID
    # --------------------------------------------------------

    df["signal_id"] = (
        df["ConstellationType"].astype(str)
        + "_"
        + df["Svid"].astype(str)
        + "_"
        + df["SignalType"].astype(str)
    )

    # --------------------------------------------------------
    # SORT BY SIGNAL + TIME
    # --------------------------------------------------------

    df = df.sort_values(
        ["signal_id", "utcTimeMillis"]
    ).reset_index(drop=True)

    # --------------------------------------------------------
    # INTERPOLATION
    # --------------------------------------------------------

    numeric_cols = [
        "Cn0DbHz",
        "RawPseudorangeMeters",
        "PseudorangeRateMetersPerSecond",
        "AccumulatedDeltaRangeMeters",
        "SvElevationDegrees",
        "SvAzimuthDegrees"
    ]

    df[numeric_cols] = (
        df.groupby("signal_id")[numeric_cols]
          .transform(
              lambda x: x.interpolate(
                  method="linear",
                  limit=2,
                  limit_direction="both"
              )
          )
    )

    df = df.dropna(
        subset=numeric_cols
    ).reset_index(drop=True)

    # --------------------------------------------------------
    # C/N0 CHANGE
    # --------------------------------------------------------

    df["Cn0_change"] = (
        df.groupby("signal_id")["Cn0DbHz"].diff()
    )

    # --------------------------------------------------------
    # C/N0 ROLLING MEAN
    # --------------------------------------------------------

    df["Cn0_rolling_mean"] = (
        df.groupby("signal_id")["Cn0DbHz"]
          .transform(
              lambda x: x.rolling(
                  window=5,
                  min_periods=1
              ).mean()
          )
    )

    # --------------------------------------------------------
    # C/N0 TREND
    # --------------------------------------------------------

    df["Cn0_trend"] = (
        df.groupby("signal_id")["Cn0DbHz"]
          .transform(
              lambda x: x.diff(4)
          )
    )

    # --------------------------------------------------------
    # C/N0 TREND MEAN
    # --------------------------------------------------------

    df["Cn0_trend_mean"] = (
        df.groupby("signal_id")["Cn0_change"]
          .transform(
              lambda x: x.rolling(
                  window=3,
                  min_periods=1
              ).mean()
          )
    )

    # --------------------------------------------------------
    # UP / DOWN
    # --------------------------------------------------------

    df["is_up"] = (
        df["Cn0_trend_mean"] > 0
    )

    df["up_streak"] = (
        df.groupby("signal_id")["is_up"]
          .transform(
              lambda x:
                  x.astype(int)
                   .groupby((x != x.shift()).cumsum())
                   .cumsum()
          )
    )

    df["is_down"] = (
        df["Cn0_trend_mean"] < 0
    )

    df["down_streak"] = (
        df.groupby("signal_id")["is_down"]
          .transform(
              lambda x:
                  x.astype(int)
                   .groupby((x != x.shift()).cumsum())
                   .cumsum()
          )
    )

    # --------------------------------------------------------
    # NEW FEATURE 1: ROLLING STD
    # --------------------------------------------------------

    df["Cn0_rolling_std"] = (
        df.groupby("signal_id")["Cn0DbHz"]
          .transform(
              lambda x: x.rolling(
                  window=5,
                  min_periods=2
              ).std()
          )
    )

    # --------------------------------------------------------
    # NEW FEATURE 2: SLOPE
    # --------------------------------------------------------

    df["Cn0_slope"] = (
        df.groupby("signal_id")["Cn0DbHz"]
          .transform(
              lambda x: x.diff(4) / 4
          )
    )

    # --------------------------------------------------------
    # NEW FEATURE 3: ACCELERATION
    # --------------------------------------------------------

    df["Cn0_acceleration"] = (
        df.groupby("signal_id")["Cn0_slope"].diff()
    )

    # --------------------------------------------------------
    # TIME GAP
    # --------------------------------------------------------

    df["time_gap"] = (
        df.groupby("signal_id")["utcTimeMillis"].diff()
    )

    # --------------------------------------------------------
    # PSEUDORANGE CHANGE
    # --------------------------------------------------------

    df["Pseudorange_change"] = (
        df.groupby("signal_id")[
            "RawPseudorangeMeters"
        ].diff()
    )

    # --------------------------------------------------------
    # PSEUDORANGE RATE CHANGE
    # --------------------------------------------------------

    df["PseudorangeRate_change"] = (
        df.groupby("signal_id")[
            "PseudorangeRateMetersPerSecond"
        ].diff()
    )

    # --------------------------------------------------------
    # ELEVATION CHANGE
    # --------------------------------------------------------

    df["Elevation_change"] = (
        df.groupby("signal_id")[
            "SvElevationDegrees"
        ].diff()
    )

    # --------------------------------------------------------
    # CIRCULAR AZIMUTH CHANGE
    # --------------------------------------------------------

    df["Azimuth_change"] = (
        (
            df.groupby("signal_id")[
                "SvAzimuthDegrees"
            ].diff() + 180
        ) % 360
    ) - 180

    df["Azimuth_change"] = (
        df["Azimuth_change"].clip(-180, 180)
    )

    # --------------------------------------------------------
    # NEW FEATURE 4: MIN
    # --------------------------------------------------------

    df["Cn0_min_5"] = (
        df.groupby("signal_id")["Cn0DbHz"]
          .transform(
              lambda x: x.rolling(
                  window=5,
                  min_periods=1
              ).min()
          )
    )

    # --------------------------------------------------------
    # NEW FEATURE 5: MAX
    # --------------------------------------------------------

    df["Cn0_max_5"] = (
        df.groupby("signal_id")["Cn0DbHz"]
          .transform(
              lambda x: x.rolling(
                  window=5,
                  min_periods=1
              ).max()
          )
    )

    # --------------------------------------------------------
    # NEW FEATURE 6: RANGE
    # --------------------------------------------------------

    df["Cn0_range_5"] = (
        df["Cn0_max_5"] - df["Cn0_min_5"]
    )

    # --------------------------------------------------------
    # NEW FEATURE 7: NEGATIVE COUNT
    # --------------------------------------------------------

    df["negative_change_count_10"] = (
        df.groupby("signal_id")["Cn0_change"]
          .transform(
              lambda x:
                  (x < 0)
                  .rolling(
                      window=10,
                      min_periods=1
                  )
                  .sum()
          )
    )

    # --------------------------------------------------------
    # NEW FEATURE 8: POSITIVE COUNT
    # --------------------------------------------------------

    df["positive_change_count_10"] = (
        df.groupby("signal_id")["Cn0_change"]
          .transform(
              lambda x:
                  (x > 0)
                  .rolling(
                      window=10,
                      min_periods=1
                  )
                  .sum()
          )
    )

    # REMOVE ROWS WHERE 20 FEATURES ARE INCOMPLETE

    df = df.dropna(
        subset=input_features
    ).reset_index(drop=True)


    # VERIFY

    missing_features = [
        f for f in input_features
        if f not in df.columns
    ]

    assert not missing_features, \
        f"Missing features: {missing_features}"

    remaining_nan = (
        df[input_features]
        .isna()
        .sum()
        .sum()
    )

    assert remaining_nan == 0, \
        f"Remaining NaNs: {remaining_nan}"

    output_name = f"gnss_{idx:03d}.parquet"

    output_path = os.path.join(
        corrected_dir,
        output_name
    )

    df.to_parquet(
        output_path,
        index=False
    )

    print(
        f"Saved {output_name} | "
        f"Rows: {len(df)} | "
        f"Columns: {len(df.columns)}"
    )

print("\n========================================")
print("DONE — ALL 196 FILES PROCESSED")
print("========================================")
```

```python
signal_weak = (
    (train_sample["Cn0DbHz"] < T["cn0_weak"]) &
    (train_sample["Cn0_min_5"] < T["cn0_min_weak"]) &
    (train_sample["Cn0_max_5"] < T["cn0_max_weak"])
)

signal_strong = (
    (train_sample["Cn0DbHz"] >= T["cn0_strong"]) &
    (train_sample["Cn0_min_5"] >= T["cn0_min_strong"]) &
    (train_sample["Cn0_max_5"] >= T["cn0_max_strong"])
)



direction_falling = (
    (train_sample["Cn0_change"] <= T["change_down"]) &
    (train_sample["Cn0_trend"] <= T["trend_down"]) &
    (train_sample["Cn0_trend_mean"] <= T["trend_mean_down"]) &
    (
        (train_sample["Cn0_slope"] <= T["slope_down"]) |
        (train_sample["Cn0_acceleration"] <= T["accel_down"])
    )
)

direction_rising = (
    (train_sample["Cn0_change"] >= T["change_up"]) &
    (train_sample["Cn0_trend"] >= T["trend_up"]) &
    (train_sample["Cn0_trend_mean"] >= T["trend_mean_up"]) &
    (
        (train_sample["Cn0_slope"] >= T["slope_up"]) |
        (train_sample["Cn0_acceleration"] >= T["accel_up"])
    )
)

direction_stable = (
    (train_sample["Cn0_change"].abs() < abs(T["change_up"])) &
    (train_sample["Cn0_trend"].abs() < max(
        abs(T["trend_down"]),
        abs(T["trend_up"])
    )) &
    (train_sample["Cn0_trend_mean"].abs() < max(
        abs(T["trend_mean_down"]),
        abs(T["trend_mean_up"])
    )) &
    (train_sample["Cn0_slope"].abs() < max(
        abs(T["slope_down"]),
        abs(T["slope_up"])
    )) &
    (train_sample["Cn0_acceleration"].abs() < max(
        abs(T["accel_down"]),
        abs(T["accel_up"])
    ))
)


persistent_decline = (
    (train_sample["down_streak"] >= T["down_strong"]) &
    (train_sample["negative_change_count_10"] >= T["negative_high"]) &
    (train_sample["up_streak"] <= T["up_none"]) &
    (train_sample["positive_change_count_10"] <= T["positive_medium"])
)

persistent_recovery = (
    (train_sample["up_streak"] >= T["up_strong"]) &
    (train_sample["positive_change_count_10"] >= T["positive_high"]) &
    (train_sample["down_streak"] <= T["down_none"]) &
    (train_sample["negative_change_count_10"] <= T["negative_medium"])
)




instability_high = (
    (train_sample["Cn0_rolling_std"] >= T["std_high"]) &
    (train_sample["Cn0_range_5"] >= T["range_high"])
)

instability_low = (
    (train_sample["Cn0_rolling_std"] <= T["std_low"]) &
    (train_sample["Cn0_range_5"] <= T["range_low"])
)




pseudorange_abnormal = (
    (train_sample["PseudorangeRateMetersPerSecond"] < T["pr_rate_low"]) |
    (train_sample["PseudorangeRateMetersPerSecond"] > T["pr_rate_high"]) |
    (train_sample["PseudorangeRate_change"] < T["pr_change_low"]) |
    (train_sample["PseudorangeRate_change"] > T["pr_change_high"])
)

pseudorange_normal = ~pseudorange_abnormal


# 6. GEOMETRY

azimuth_valid = (
    train_sample["SvAzimuthDegrees"].between(
        T["azimuth_min"],
        T["azimuth_max"]
    )
)

geometry_unfavorable = (
    (~azimuth_valid) |
    (train_sample["SvElevationDegrees"] < T["elevation_bad"]) |
    (
        (train_sample["SvElevationDegrees"] < T["elevation_bad"]) &
        (
            (train_sample["Elevation_change"] <= T["elevation_change_down"]) |
            (train_sample["Azimuth_change"] <= T["azimuth_change_down"]) |
            (train_sample["Azimuth_change"] >= T["azimuth_change_up"])
        )
    )
)

geometry_favorable = (
    azimuth_valid &
    (train_sample["SvElevationDegrees"] >= T["elevation_good"]) &
    (train_sample["Elevation_change"] > T["elevation_change_down"]) &
    (train_sample["Elevation_change"] < T["elevation_change_up"]) &
    (train_sample["Azimuth_change"] > T["azimuth_change_down"]) &
    (train_sample["Azimuth_change"] < T["azimuth_change_up"])
)


```

```python
# ============================================================
# STEP A — LABEL ALL 196 FILES
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

corrected_dir = Path("/content/gnss_features_v2")
labeled_dir   = Path("/content/gnss_labeled_final")
labeled_dir.mkdir(parents=True, exist_ok=True)

# Columns to load and save
COLUMNS_TO_KEEP = ALL_FEATURES + ["signal_id", "utcTimeMillis"]

# Thresholds
T = {
    "cn0_weak": 28.9,
    "cn0_min_weak": 26.5584,
    "cn0_max_weak": 30.9841,
    "change_down": -0.7000,
    "trend_down": -1.3577,
    "trend_mean_down": -0.4000,
    "slope_down": -0.3394,
    "accel_down": -0.2816,
    "change_up": 0.6914,
    "trend_up": 1.3000,
    "trend_mean_up": 0.3877,
    "slope_up": 0.3250,
    "accel_up": 0.2750,
}


def label_dataframe(df):
    """Apply 8-feature labeling rules."""

    signal_weak = (
        (df["Cn0DbHz"]   < T["cn0_weak"]) &
        (df["Cn0_min_5"] < T["cn0_min_weak"]) &
        (df["Cn0_max_5"] < T["cn0_max_weak"])
    )

    direction_falling = (
        (df["Cn0_change"]     <= T["change_down"]) &
        (df["Cn0_trend"]      <= T["trend_down"]) &
        (df["Cn0_trend_mean"] <= T["trend_mean_down"]) &
        (
            (df["Cn0_slope"] <= T["slope_down"]) |
            (df["Cn0_acceleration"] <= T["accel_down"])
        )
    )

    direction_rising = (
        (df["Cn0_change"]     >= T["change_up"]) &
        (df["Cn0_trend"]      >= T["trend_up"]) &
        (df["Cn0_trend_mean"] >= T["trend_mean_up"]) &
        (
            (df["Cn0_slope"] >= T["slope_up"]) |
            (df["Cn0_acceleration"] >= T["accel_up"])
        )
    )

    BAD       = signal_weak & direction_falling
    DEGRADING = direction_falling & (~signal_weak)
    RECOVERY  = direction_rising
    GOOD      = (~direction_falling) & (~direction_rising)

    df["gnss_state"] = np.select(
        [BAD, DEGRADING, RECOVERY, GOOD],
        ["BAD", "DEGRADING", "RECOVERY", "GOOD"],
        default="GOOD"
    )
    return df


# ------------------------------------------------------------
# Loop over 196 files
# ------------------------------------------------------------
total_rows = 0
total_counts = {"BAD": 0, "DEGRADING": 0, "RECOVERY": 0, "GOOD": 0}

for i in range(1, 197):

    in_path  = corrected_dir / f"gnss_{i:03d}.parquet"
    out_path = labeled_dir   / f"gnss_{i:03d}.parquet"

    df = pd.read_parquet(in_path, columns=COLUMNS_TO_KEEP)
    df = label_dataframe(df)
    df.to_parquet(out_path, index=False)

    counts = df["gnss_state"].value_counts()
    for state in total_counts:
        total_counts[state] += int(counts.get(state, 0))

    total_rows += len(df)

    if i % 25 == 0 or i == 196:
        print(f"[{i}/196] done")


# ------------------------------------------------------------
# Final distribution
# ------------------------------------------------------------
print("\n====================================================")
print("FINAL LABELS — ALL 196 FILES")
print("====================================================")
for state in ["BAD", "DEGRADING", "RECOVERY", "GOOD"]:
    pct = total_counts[state] / total_rows * 100
    print(f"{state:12s}: {total_counts[state]:>10,} ({pct:.2f}%)")

print(f"\nTotal rows: {total_rows:,}")
print(f"Saved to : {labeled_dir}")
```
