```python
import os
import pandas as pd
import numpy as np

labeled_dir = "/content/gnss_labeled"

input_features = [
    "Cn0DbHz",
    "Cn0_change",
    "Cn0_trend",
    "Cn0_trend_mean",
    "down_streak",
    "up_streak",
    "PseudorangeRateMetersPerSecond",
    "PseudorangeRate_change",
    "SvElevationDegrees",
    "Elevation_change",
    "SvAzimuthDegrees",
    "Azimuth_change"
]

sequence_length = 10
prediction_horizon = 5

X = []
y = []

labeled_files = sorted([
    f for f in os.listdir(labeled_dir)
    if f.endswith("_labeled.parquet")
])

for file in labeled_files:

    df = pd.read_parquet(
        os.path.join(labeled_dir, file)
    )

    for signal_id, group in df.groupby("signal_id"):

        group = (
            group
            .sort_values("utcTimeMillis")
            .reset_index(drop=True)
        )

        group = group.dropna(
            subset=input_features
        ).reset_index(drop=True)

        if len(group) < sequence_length + prediction_horizon:
            continue

        features = group[input_features].values
        states = group["state"].values

        for i in range(
            len(group) - sequence_length - prediction_horizon + 1
        ):
            X.append(
                features[i:i + sequence_length]
            )

            target_index = (
                i
                + sequence_length
                + prediction_horizon
                - 1
            )

            y.append(states[target_index])

X = np.array(X)
y = np.array(y)

print("X shape:", X.shape)
print("y shape:", y.shape)
print("\nState distribution:")
print(pd.Series(y).value_counts())
```

```python
import os
import pandas as pd
import numpy as np

labeled_dir = "/content/gnss_labeled"

files = sorted([
    f for f in os.listdir(labeled_dir)
    if f.endswith("_labeled.parquet")
])

print("Total labeled files:", len(files))

# --------------------------------------------------
# Expected columns
# --------------------------------------------------

expected_features = [
    "utcTimeMillis",
    "Svid",
    "Cn0DbHz",
    "PseudorangeRateMetersPerSecond",
    "AccumulatedDeltaRangeMeters",
    "ConstellationType",
    "RawPseudorangeMeters",
    "SignalType",
    "SvElevationDegrees",
    "SvAzimuthDegrees",
    "signal_id",
    "Cn0_change",
    "time_gap",
    "Pseudorange_change",
    "PseudorangeRate_change",
    "Elevation_change",
    "Azimuth_change",
    "Cn0_rolling_mean",
    "Cn0_trend",
    "Cn0_trend_mean",
    "is_down",
    "down_streak",
    "is_up",
    "up_streak",
    "state"
]

expected_states = {
    "GOOD",
    "DEGRADING",
    "BAD",
    "RECOVERY"
}

# --------------------------------------------------
# Overall counters
# --------------------------------------------------

total_rows = 0
total_state_counts = {
    "GOOD": 0,
    "DEGRADING": 0,
    "BAD": 0,
    "RECOVERY": 0
}

files_with_wrong_columns = []
files_with_bad_states = []
files_with_bad_signal_id = []
files_with_time_order_problem = []
files_with_nan = []

# --------------------------------------------------
# Audit each file
# --------------------------------------------------

for idx, file in enumerate(files, 1):

    path = os.path.join(labeled_dir, file)
    df = pd.read_parquet(path)

    total_rows += len(df)

    # 1. Column check
    missing_cols = [
        col for col in expected_features
        if col not in df.columns
    ]

    if missing_cols:
        files_with_wrong_columns.append(
            (file, missing_cols)
        )

    # 2. State check
    state_values = set(df["state"].dropna().unique())

    if not state_values.issubset(expected_states):
        files_with_bad_states.append(
            (file, sorted(state_values - expected_states))
        )

    # Count states
    counts = df["state"].value_counts()

    for state in expected_states:
        total_state_counts[state] += int(
            counts.get(state, 0)
        )

    # 3. signal_id check
    if df["signal_id"].isna().any():
        files_with_bad_signal_id.append(file)

    # 4. Time ordering inside every signal
    bad_order = False

    for signal_id, group in df.groupby("signal_id"):
        times = group["utcTimeMillis"].values

        if len(times) > 1:
            if not np.all(times[1:] >= times[:-1]):
                bad_order = True
                break

    if bad_order:
        files_with_time_order_problem.append(file)

    # 5. NaN check in important columns
    important_cols = [
        "Cn0DbHz",
        "Cn0_change",
        "time_gap",
        "Pseudorange_change",
        "PseudorangeRate_change",
        "Elevation_change",
        "Azimuth_change",
        "Cn0_trend",
        "Cn0_trend_mean",
        "state"
    ]

    if df[important_cols].isna().any().any():
        files_with_nan.append(file)

    if idx % 20 == 0 or idx == len(files):
        print(f"Checked {idx}/{len(files)} files")

# --------------------------------------------------
# Final report
# --------------------------------------------------

print("\n" + "="*60)
print("FINAL 196-FILE LABEL AUDIT")
print("="*60)

print("\nFiles checked:", len(files))
print("Total rows:", total_rows)

print("\nState distribution:")
for state, count in total_state_counts.items():
    percentage = (count / total_rows) * 100
    print(f"{state:10s}: {count:10d} ({percentage:.2f}%)")

print("\nProblems found:")
print("Wrong columns:", len(files_with_wrong_columns))
print("Bad states:", len(files_with_bad_states))
print("Bad signal_id:", len(files_with_bad_signal_id))
print("Time-order problems:", len(files_with_time_order_problem))
print("Files with NaN in important columns:", len(files_with_nan))

if files_with_wrong_columns:
    print("\nColumn problems:")
    print(files_with_wrong_columns[:10])

if files_with_bad_states:
    print("\nState problems:")
    print(files_with_bad_states[:10])

if files_with_bad_signal_id:
    print("\nSignal ID problems:")
    print(files_with_bad_signal_id[:10])

if files_with_time_order_problem:
    print("\nTime-order problems:")
    print(files_with_time_order_problem[:10])

if files_with_nan:
    print("\nNaN files:")
    print(files_with_nan[:10])
```
