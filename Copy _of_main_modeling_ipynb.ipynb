```python
import os
import gc
import zipfile
import pandas as pd

# Folder where cleaned files will be stored
output_dir = "/content/gnss_cleaned"
os.makedirs(output_dir, exist_ok=True)

with zipfile.ZipFile(zip_path, "r") as z:

    for i, file in enumerate(gnss_files, start=1):

        print(f"\nProcessing {i}/{len(gnss_files)}")
        print(file)

        # Read only the columns we need
        df = pd.read_csv(
            z.open(file),
            usecols=features
        )

        # Remove rows with missing values
        df = df.dropna()

        # Sort by satellite and time
        df = df.sort_values(
            ["Svid", "utcTimeMillis"]
        ).reset_index(drop=True)

        # Save cleaned file
        output_file = os.path.join(
            output_dir,
            f"gnss_{i:03d}.parquet"
        )

        df.to_parquet(
            output_file,
            index=False
        )

        print("Cleaned shape:", df.shape)
        print("Saved:", output_file)

        # Free RAM before next file
        del df
        gc.collect()

print("\n✅ All GNSS files cleaned and saved.")
```
