```python
import zipfile

zip_path = "/content/drive/MyDrive/Gnns_kaggle_2324/smartphone-decimeter-2023.zip"

with zipfile.ZipFile(zip_path, "r") as z:
    gnss_files = [
        f for f in z.namelist()
        if "/train/" in f and f.endswith("device_gnss.csv")
    ]

print("Training GNSS files:", len(gnss_files))
print("\nFirst 10:")
for f in gnss_files[:10]:
    print(f)
```

```python
  import pandas as pd

  sample_file = gnss_files[0]

  with zipfile.ZipFile(zip_path) as z:
      with z.open(sample_file) as f:
          gnss_df = pd.read_csv(f)

  print("File:", sample_file)
  print("Shape:", gnss_df.shape)

  display(gnss_df.head())
  print("\nData types:")
  display(gnss_df.dtypes)

  print("\nMissing values:")
  display(gnss_df.isna().sum().sort_values(ascending=False))

```

```python
satellites_count = (gnss_df[["ConstellationType","Svid","SignalType"]].drop_duplicates().sort_values(by=["ConstellationType","Svid","SignalType"]))
display(satellites_count)
```

```python
chosen_constellitation = 1
chosen_svid = 8
chosen_signal = "GPS_L1_CA"
print(chosen_constellitation)
print(chosen_svid)
print(chosen_signal)
```

```python
satellites_df = gnss_df[
    (gnss_df["ConstellationType"] == chosen_constellitation) &
    (gnss_df["Svid"] == chosen_svid) &
    (gnss_df["SignalType"] == chosen_signal)
].sort_values(by="utcTimeMillis").copy()

print(len(satellites_df))
```

```python
time_differences = satellites_df["utcTimeMillis"].diff().dropna()
display(time_differences.value_counts().sort_index())
```

```python

  satellites_df["time_gap_ms"] = satellites_df["utcTimeMillis"].diff()

  gaps = satellites_df[satellites_df["time_gap_ms"] > 1000]

  display(gaps[["utcTimeMillis", "time_gap_ms", "Cn0DbHz"]])

```

```python
  important_columns = [
      "Cn0DbHz",
      "SvElevationDegrees",
      "SvAzimuthDegrees",
      "PseudorangeRateMetersPerSecond",
      "MultipathIndicator"
  ]

  display(satellites_df[important_columns].isna().sum())

```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 5))

plt.plot(
    satellites_df["utcTimeMillis"],
    satellites_df["Cn0DbHz"],
    marker=".",
    markersize=3
)

plt.title("GPS Satellite 8: Signal Strength Over Time")
plt.xlabel("Time")
plt.ylabel("C/N0 (dB-Hz)")
plt.grid(True)
plt.show()
```
